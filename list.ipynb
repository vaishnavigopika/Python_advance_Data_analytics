```python
mini=10000
pin=input("enter your PIN")
ch='y'
if (len(pin)==4):
    while(ch=='y'):
        print("1.Withdraw \n 2.Enquiry/n3.Exit")
        n=int(input("Enter your choice "))
        if(n==1):
            amt=int(input("Enter the amount to be withdrawen"))
            mini=mini-amt
            
        elif(n==2):
            print("Your current Balance is ",mini)
            
        elif n==3:
            ch=input("y/n")
        
        else:
            print("CHOICE NOT FOUND")
        
else:
    print("INVALID PIN")
```

```python
import csv
data = {}
with open('DS.csv', 'r') as file:
    csvFile = csv.reader(file)
    headers = next(csvFile) 
    for h in headers:
        data[h] = []
    for row in csvFile:
        for i in range(len(headers)):
            data[headers[i]].append(row[i])

print(data)
```
